```python
import os
import hashlib
from PIL import Image
import imagehash
import pandas as pd
from collections import defaultdict
```

```python
IMAGE_FOLDER = r"f:\Samir\Projects\SnakeSense\Filtered_Train"
```

```python
md5_hashes = defaultdict(list)

for filename in os.listdir(IMAGE_FOLDER):

    if not filename.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    filepath = os.path.join(IMAGE_FOLDER, filename)

    with open(filepath, "rb") as f:
        file_hash = hashlib.md5(f.read()).hexdigest()

    md5_hashes[file_hash].append(filename)
```

```python
exact_duplicates = {
    h: files
    for h, files in md5_hashes.items()
    if len(files) > 1
}

print(f"Exact duplicate groups: {len(exact_duplicates)}")

total_duplicate_images = sum(len(v)-1 for v in exact_duplicates.values())

print(f"Duplicate images: {total_duplicate_images}")
```

```python
phashes = {}

for filename in os.listdir(IMAGE_FOLDER):

    if not filename.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    filepath = os.path.join(IMAGE_FOLDER, filename)

    try:
        img = Image.open(filepath)

        phashes[filename] = imagehash.phash(img)

    except:
        pass
```

```python
near_duplicates = []

filenames = list(phashes.keys())

threshold = 5

for i in range(len(filenames)):

    for j in range(i + 1, len(filenames)):

        distance = phashes[filenames[i]] - phashes[filenames[j]]

        if distance <= threshold:

            near_duplicates.append({

                "Image1": filenames[i],
                "Image2": filenames[j],
                "Distance": distance

            })
near_df = pd.DataFrame(near_duplicates)
```

```python
print("="*50)
print("DUPLICATE DETECTION REPORT")
print("="*50)

print(f"Total Images           : {len(os.listdir(IMAGE_FOLDER))}")
print(f"Exact Duplicate Groups : {len(exact_duplicates)}")
print(f"Exact Duplicate Images : {total_duplicate_images}")
print(f"Near Duplicate Pairs   : {len(near_df)}")
```
