```python
OUTPUT_DIR = "data/Processed_Train/images"

os.makedirs(OUTPUT_DIR, exist_ok=True)

print("Folder created.")
```

```python
# Input
IMAGE_FOLDER = "data/Filtered_Train"
OUTPUT_FOLDER = "data/Processed_Train/images"
```

```python
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

processed_rows = []

PADDING = 0.15  # 10% padding

for _, row in filtered_species.iterrows():

    image_path = os.path.join(IMAGE_FOLDER, row["filename"])

    try:
        image = Image.open(image_path).convert("RGB")

        img_width, img_height = image.size

        xmin = int(row["xmin"])
        ymin = int(row["ymin"])
        xmax = int(row["xmax"])
        ymax = int(row["ymax"])

        box_width = xmax - xmin
        box_height = ymax - ymin

        pad_x = int(box_width * PADDING)
        pad_y = int(box_height * PADDING)

        xmin = max(0, xmin - pad_x)
        ymin = max(0, ymin - pad_y)
        xmax = min(img_width, xmax + pad_x)
        ymax = min(img_height, ymax + pad_y)

        cropped = image.crop((xmin, ymin, xmax, ymax))

        save_path = os.path.join(OUTPUT_FOLDER, row["filename"])
        cropped.save(save_path)

        processed_rows.append({
            "filename": row["filename"],
            "class": row["class"],
            "Label": row["Label"],
            "width": cropped.size[0],
            "height": cropped.size[1]
        })

    except Exception as e:
        print(f"Error: {row['filename']} -> {e}")
```

```python
processed_annotations = pd.DataFrame(processed_rows)

processed_annotations.to_csv(
    "data/Processed_Train/processed_annotations.csv",
    index=False
)

print(f"Successfully processed {len(processed_annotations)} images.")
```

## Test data

```python
import shutil
```

```python
SOURCE_FOLDER = r"data/raw/test"
CSV_FILE = os.path.join(SOURCE_FOLDER, "_annotations.csv")
DEST_FOLDER = r"data/Processed_Images/Processed_test"
```

```python
TARGET_SPECIES = {
    "naja naja",
    "bungarus caeruleus",
    "daboia russelii",
    "ptyas korros",
    "ophiophagus hannah",
    "bungarus fasciatus",
    "chrysopelea ornata",
    "ptyas mucosa",
    "xenochrophis piscator",
    "amphiesma stolatum",
    "dendrelaphis pictus",
    "ahaetulla prasina",
    "psammodynastes pulverulentus",
    "trimeresurus albolabris",
    "trimeresurus purpureomaculatus"
}
```

```python
def analyze_test_data():
    process_test = pd.read_csv(CSV_FILE)
    
    # Normalize class names
    process_test["class"] = (
        process_test["class"]
            .str.strip()
            .str.lower()
    )
    # Handle synonyms
    process_test["class"] = process_test["class"].replace({
        "fowlea piscator": "xenochrophis piscator",
        "hebius stolatum": "amphiesma stolatum"
    })
    
    # Keep only target species
    filtered = process_test[process_test["class"].isin(TARGET_SPECIES)]

    # Remove duplicate filenames
    filtered = filtered.drop_duplicates(subset="filename")

    print(f"Images to copy: {len(filtered)}")
    
    
    # ===========================
    # COPY IMAGES
    # ===========================

    copied = 0
    skipped = 0
    missing = []

    os.makedirs(DEST_FOLDER, exist_ok=True)

    for filename in filtered["filename"].drop_duplicates().tolist():
        src = os.path.join(SOURCE_FOLDER, filename)
        dst = os.path.join(DEST_FOLDER, filename)

        if not os.path.exists(src):
            missing.append(filename)
            continue

        if os.path.exists(dst):
            skipped += 1
            continue

        shutil.copy2(src, dst)
        copied += 1
        
    print(f"\nCopied : {copied}")
    print(f"Skipped: {skipped}")
    print(f"Missing: {len(missing)}")

    if missing:
        print("\nMissing files:")
        for m in missing[:10]:
            print(m)
            
            
    # ===========================
    # SAVE FILTERED ANNOTATIONS
    # ===========================

    output_csv = os.path.join(DEST_FOLDER, "_annotations.csv")

    if os.path.exists(output_csv):
        print(f"{output_csv} already exists. Skipping save.")
    else:
        filtered.to_csv(output_csv, index=False)
        print(f"Saved {output_csv}")
```

```python
TEST_CSV = pd.read_csv("data/Processed_Images/Processed_test\_annotations.csv")

classes = sorted(TEST_CSV["class"].unique()) # sorts them alphabetically.

class_to_idx = { # mapping integers {"cobra": 0, "python": 1, "viper": 2}
    cls: idx
    for idx, cls in enumerate(classes) # enumerate give pairs like (0, "cobra"), (1, "python")
}

idx_to_class = { # reverse {0: "cobra", 1: "python", 2: "viper"}
    idx: cls
    for cls, idx in class_to_idx.items() 
}

```

```python
# creates a new column called Label
TEST_CSV["Label"] = TEST_CSV["class"].map(class_to_idx) #  applies the mapping to every value in the class column.
TEST_CSV.to_csv("data/Processed_Images/Processed_test_annotations.csv", index=False)
TEST_CSV
```
